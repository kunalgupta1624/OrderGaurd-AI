```python
import pandas as pd
import numpy as np

df = pd.read_csv("../dataset/orders_sample.csv")

print("Original shape:", df.shape)

df.head()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
categorical_columns = [
    'city',
    'state',
    'utm_source',
    'courier',
    'payment_method',
    'final_status'
]

for col in categorical_columns:
    df[col] = (
        df[col]
        .astype('string')
        .str.strip()
    )

df['final_status'] = df['final_status'].str.upper()

print(df[categorical_columns].head())
```

```python
print(df.isnull().sum())
```

```python
df['city'] = df['city'].fillna('UNKNOWN')

df['pincode'] = df['pincode'].astype('string')

df['pincode'] = df['pincode'].replace(
    ['<NA>', 'nan', 'None'],
    pd.NA
)

print(df.isnull().sum())
```

```python
duplicates = df[
    df.duplicated(
        subset=['order_id'],
        keep=False
    )
].sort_values('order_id')

print("Duplicate records:", len(duplicates))

duplicates.head(20)
```

```python
duplicate_summary = (
    duplicates
    .groupby('order_id')
    .agg(
        record_count=('order_id', 'size'),
        different_statuses=('final_status', 'nunique')
    )
)

print(duplicate_summary.head(20))
```

```python
df['customer_rto_rate'] = np.where(
    df['customer_prior_orders'] > 0,
    df['customer_prior_rto'] /
    df['customer_prior_orders'],
    0
)

df['customer_rto_rate'] = (
    df['customer_rto_rate'].clip(0, 1)
)
```

```python
df['customer_successful_orders'] = (
    df['customer_prior_orders'] -
    df['customer_prior_rto']
)

df['customer_successful_orders'] = (
    df['customer_successful_orders'].clip(lower=0)
)
```

```python
def calculate_address_quality(address):

    if pd.isna(address):
        return 0

    address = str(address).strip().lower()

    if address in ['', '.', 'unknown', 'na', 'null']:
        return 0

    score = 0

    if len(address) >= 15:
        score += 30

    if any(char.isdigit() for char in address):
        score += 20

    location_keywords = [
        'road', 'street', 'nagar',
        'colony', 'sector', 'block',
        'apartment', 'flat', 'house',
        'near', 'opposite', 'mandir',
        'chowk', 'village'
    ]

    if any(word in address for word in location_keywords):
        score += 30

    if len(address.split()) >= 4:
        score += 20

    return min(score, 100)
```

```python
df['address_quality_score'] = (
    df['address'].apply(calculate_address_quality)
)
```

```python
def validate_customer_name(name):

    if pd.isna(name):
        return 0

    name = str(name).strip().lower()

    suspicious_names = [
        'asdf', 'test', 'xyz',
        'qwerty', 'abc', 'unknown'
    ]

    if name in suspicious_names:
        return 0

    if len(name) < 3:
        return 0

    return 1
```

```python
df['name_valid'] = (
    df['customer_name'].apply(validate_customer_name)
)
```

```python
def validate_pincode(pincode):

    if pd.isna(pincode):
        return 0

    pincode = str(pincode).strip()

    if pincode.endswith('.0'):
        pincode = pincode[:-2]

    return int(
        len(pincode) == 6 and
        pincode.isdigit() and
        pincode[0] != '0'
    )
```

```python
df['pincode_valid'] = (
    df['pincode'].apply(validate_pincode)
)
```

```python
df['high_value_order'] = (
    df['order_value'] >= 1000
).astype(int)

df['value_per_item'] = (
    df['order_value'] /
    df['num_items'].replace(0, np.nan)
)
```

```python
df['order_datetime'] = pd.to_datetime(
    df['order_datetime'],
    errors='coerce'
)

df['order_hour'] = df['order_datetime'].dt.hour

df['order_day'] = df['order_datetime'].dt.day

df['order_weekday'] = (
    df['order_datetime'].dt.day_of_week
)

df['is_weekend'] = (
    df['order_weekday'] >= 5
).astype(int)
```

Create the ML Dataset

```python
df['final_status'] = (
    df['final_status']
    .astype(str)
    .str.strip()
    .str.upper()
)

ml_df = df[
    df['final_status'].isin(['DELIVERED', 'RTO'])
].copy()

ml_df['is_rto'] = (
    ml_df['final_status'] == 'RTO'
).astype(int)

print(ml_df['is_rto'].value_counts())
```

```python
feature_columns = [
    'order_value',
    'num_items',
    'customer_prior_orders',
    'customer_prior_rto',
    'customer_rto_rate',
    'customer_successful_orders',
    'address_quality_score',
    'name_valid',
    'pincode_valid',
    'high_value_order',
    'value_per_item',
    'order_hour',
    'order_weekday',
    'is_weekend',
    'city',
    'state',
    'courier',
    'utm_source'
]

X = ml_df[feature_columns].copy()

y = ml_df['is_rto'].copy()

print("Feature matrix:", X.shape)

print("Target matrix:", y.shape)
```

```python
ml_df.to_csv(
    "../dataset/processed_orders.csv",
    index=False
)

print("Processed dataset saved successfully!")
```
